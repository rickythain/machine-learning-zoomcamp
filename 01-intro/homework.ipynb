```python
# Q7 - Sum of weights
```

```python
X = df[ df['origin'] == 'Asia'] [ ['vehicle_weight', 'model_year'] ].head(7).to_numpy()
print(X)
```

```python
XTX = X.T.dot(X)
print(XTX)
```

```python
XTX = np.linalg.inv(XTX)
print(XTX)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = np.dot(np.dot(XTX, X.T), y)
print(w.sum())
```
